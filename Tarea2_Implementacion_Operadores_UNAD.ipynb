```python
# ==============================================================================
# ALGORITMO: EVALUACIÓN DE RENDIMIENTO DE VENTAS POR SUCURSAL
# ==============================================================================

def evaluar_sucursal(nombre_sucursal, venta_mes1, venta_mes2, venta_mes3, rendimiento_minimo=1000.0):
    """
    Función que calcula la suma y el promedio de ventas, e evalúa
    si la sucursal cumple con la meta mínima utilizando operadores
    aritméticos y relacionales.
    """
    # Operador Aritmético (+): Suma total de las tres ventas
    suma_ventas = venta_mes1 + venta_mes2 + venta_mes3

    # Operador Aritmético (/): Promedio mensual
    promedio_ventas = suma_ventas / 3.0

    # Operador Relacional (>=): Verificación lógica de cumplimiento
    cumple_meta = promedio_ventas >= rendimiento_minimo  # Valor Booleano (True/False)

    # Imprimir reporte formal por consola
    print("==================================================================")
    print("      SISTEMA DE ANALÍTICA DE VENTAS - INFORME DE SUCURSAL       ")
    print("==================================================================")
    print(f"Sucursal evaluada        : {nombre_sucursal}")
    print(f"Ventas Mes 1             : ${venta_mes1:,.2f}")
    print(f"Ventas Mes 2             : ${venta_mes2:,.2f}")
    print(f"Ventas Mes 3             : ${venta_mes3:,.2f}")
    print("------------------------------------------------------------------")
    print(f"Ventas totales (3 meses) : ${suma_ventas:,.2f}")
    print(f"Promedio mensual         : ${promedio_ventas:,.2f}")
    print(f"Meta mínima requerida    : ${rendimiento_minimo:,.2f}")
    print("------------------------------------------------------------------")
    print(f"Evaluación Relacional    : {promedio_ventas:,.2f} >= {rendimiento_minimo:,.2f} -> {cumple_meta}")

    if cumple_meta:
        print("RESULTADO                : RENDIMIENTO SATISFACTORIO")
        print("DIAGNÓSTICO              : La sucursal SUPERA la meta mínima de ventas.")
    else:
        print("RESULTADO                : RENDIMIENTO INSUFICIENTE")
        print("DIAGNÓSTICO              : La sucursal NO ALCANZA la meta mínima de ventas.")
    print("==================================================================\n")


# ------------------------------------------------------------------------------
# EJECUCIÓN AUTOMÁTICA DE CASOS DE PRUEBA (PARA EVIDENCIAS)
# ------------------------------------------------------------------------------
print(">>> EJECUTANDO CASOS DE PRUEBA OFICIALES <<<\n")

# CASO 1: Sucursal que SUPERA la meta
print("--- CASO DE PRUEBA 1 ---")
evaluar_sucursal(
    nombre_sucursal="Norte",
    venta_mes1=1200.0,
    venta_mes2=1500.0,
    venta_mes3=1100.0
)

# CASO 2: Sucursal que NO ALCANZA la meta
print("--- CASO DE PRUEBA 2 ---")
evaluar_sucursal(
    nombre_sucursal="Centro",
    venta_mes1=800.0,
    venta_mes2=900.0,
    venta_mes3=850.0
)
```
