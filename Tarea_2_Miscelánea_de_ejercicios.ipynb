```python
import numpy as np

print("=" * 50)
print("EJERCICIO 1: OPERACIONES CON MATRICES")
print("=" * 50)
A = np.array([[5, 6], [7, 8]])
B = np.array([[5, 7], [6, 7]])

print("A + B:\n", A + B)
print("\n2A - 3B:\n", 2 * A - 3 * B)
print("\nA x B:\n", np.dot(A, B))
det_A = np.linalg.det(A)
print(f"\ndet(A): {det_A:.2f}")
print("¿Es invertible?:", "Sí" if round(det_A, 4) != 0 else "No")

print("\n" + "=" * 50)
print("EJERCICIO 2: SISTEMA DE ECUACIONES Y CRAMER")
print("=" * 50)
M = np.array([[2, 1, -1], [3, -2, 1], [1, 4, -3]])
b = np.array([5, -4, 3])

det_M = np.linalg.det(M)
print(f"det(M): {det_M:.2f}")

Mx = M.copy(); Mx[:, 0] = b
My = M.copy(); My[:, 1] = b
Mz = M.copy(); Mz[:, 2] = b

print(f"det(Mx): {np.linalg.det(Mx):.2f}")
print(f"det(My): {np.linalg.det(My):.2f}")
print(f"det(Mz): {np.linalg.det(Mz):.2f}")
if abs(det_M) < 1e-9:
    print("Conclusión: det(M) = 0. El sistema no tiene solución única (es inconsistente).")

print("\n" + "=" * 50)
print("EJERCICIO 3: VALORES Y VECTORES PROPIOS")
print("=" * 50)
C = np.array([[7, 8], [5, 6]])
vals, vecs = np.linalg.eig(C)

print("Valores propios:", vals)
print("Vectores propios (columnas normalizadas):\n", vecs)
```

```python
import matplotlib.pyplot as plt
import numpy as np

# --- EJERCICIO 4 ---
# Definición de la función lineal
def f_lineal(x):
    return 2 * x + 3

# Evaluación puntual
puntos_x = [2, -1, 0]
evaluaciones = {x: f_lineal(x) for x in puntos_x}
print("Evaluaciones:", evaluaciones)

# Generación de la gráfica
x = np.linspace(-5, 5, 100)
y = f_lineal(x)

plt.figure(figsize=(7, 4))
plt.plot(x, y, label='f(x) = 2x + 3', color='blue')
plt.scatter(puntos_x, [f_lineal(px) for px in puntos_x], color='red', zorder=5)
plt.axhline(0, color='black', linewidth=0.8)
plt.axvline(0, color='black', linewidth=0.8)
plt.grid(True, linestyle='--', alpha=0.6)
plt.title('Ejercicio 4: Gráfica de Función Lineal')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.legend()
plt.show()
```

```python
import sympy as sp

# --- EJERCICIO 6 ---
# Definición de la variable simbólica y la ecuación
x = sp.Symbol('x')
ecuacion = sp.Eq(x**2 - 5*x + 6, 0)

# Resolución simbólica mediante factorización
soluciones = sp.solve(ecuacion, x)
print(f"Las soluciones exactas de la ecuación no lineal son: {soluciones}")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats
from sklearn.cluster import KMeans

# =========================================================
# EJERCICIOS 7 Y 8: ESTADÍSTICA DESCRIPTIVA (EDADES)
# =========================================================
edades = np.array([25, 28, 30, 30, 30, 32, 35, 40, 45, 50])

media = np.mean(edades)
mediana = np.median(edades)
moda = stats.mode(edades).mode
rango = np.ptp(edades)
varianza = np.var(edades, ddof=1)
desv_est = np.std(edades, ddof=1)
cv = (desv_est / media) * 100

print("=" * 50)
print("EJERCICIOS 7 Y 8: MEDIDAS ESTADÍSTICAS DE EDADES")
print("=" * 50)
print(f"Media Aritmética:     {media:.2f} años")
print(f"Mediana:              {mediana:.2f} años")
print(f"Moda:                 {moda} años")
print(f"Rango:                {rango} años")
print(f"Varianza Muestral:    {varianza:.2f} años²")
print(f"Desviación Estándar:  {desv_est:.2f} años")
print(f"Coef. de Variación:   {cv:.2f}%\n")

# =========================================================
# EJERCICIO 9: VISUALIZACIÓN DE VENTAS MENSUALES
# =========================================================
meses = [f"Mes {i}" for i in range(1, 13)]
ventas = [120, 135, 150, 140, 160, 155, 170, 180, 165, 175, 190, 210]

fig, ax = plt.subplots(1, 2, figsize=(13, 4.5))

# Gráfico de Barras
ax[0].bar(meses, ventas, color='skyblue', edgecolor='black')
ax[0].set_title("Ejercicio 9: Ventas Mensuales por Mes", fontweight='bold')
ax[0].set_ylabel("Ventas (miles USD)")
ax[0].tick_params(axis='x', rotation=45)
ax[0].grid(axis='y', linestyle='--', alpha=0.7)

# Histograma de Frecuencias
ax[1].hist(ventas, bins=5, color='coral', edgecolor='black')
ax[1].set_title("Ejercicio 9: Distribución de Frecuencia de Ventas", fontweight='bold')
ax[1].set_xlabel("Rango de Ventas (miles USD)")
ax[1].set_ylabel("Frecuencia (N° de meses)")
ax[1].grid(True, linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()

# =========================================================
# EJERCICIO 10: CLUSTERING K-MEANS 2D
# =========================================================
X = np.array([
    [150, 50], [180, 60], [160, 40], [200, 70],
    [400, 300], [450, 350], [420, 310], [500, 380]
])

kmeans = KMeans(n_clusters=2, random_state=42, n_init=10).fit(X)

plt.figure(figsize=(8, 4.5))
plt.scatter(X[:, 0], X[:, 1], c=kmeans.labels_, cmap='viridis', s=100, label='Clientes', zorder=5)
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1], c='red', marker='X', s=200, label='Centroides', zorder=6)
plt.title("Ejercicio 10: Clasificación de Clientes (K-means)", fontweight='bold')
plt.xlabel("Gasto en Alimentación (USD)")
plt.ylabel("Gasto en Ocio (USD)")
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

# =========================================================
# EJERCICIO 11: OPTIMIZACIÓN DE INVENTARIOS (REGIÓN FACTIBLE)
# =========================================================
# Ecuación Presupuesto: 50x + 30y <= 5000 -> y = (5000 - 50x) / 30
# Ecuación Almacén:     x + y <= 150       -> y = 150 - x

x = np.linspace(0, 160, 400)
y_presupuesto = (5000 - 50 * x) / 30
y_almacen = 150 - x

plt.figure(figsize=(8, 5))
plt.plot(x, y_presupuesto, label='50x + 30y <= 5000 (Presupuesto)', color='blue', linewidth=2)
plt.plot(x, y_almacen, label='x + y <= 150 (Almacenamiento)', color='orange', linewidth=2)

# Sombreado de la Región Factible
y_factible = np.minimum(np.maximum(0, y_presupuesto), np.maximum(0, y_almacen))
plt.fill_between(x, 0, y_factible, where=(x >= 0) & (y_factible >= 0), color='gray', alpha=0.3, label='Región Factible')

# Intersección óptima (x=25, y=125)
plt.scatter(25, 125, color='red', s=120, zorder=5, label='Punto Óptimo (25, 125)')
plt.annotate(' P(25, 125)', (25 + 2, 125), fontsize=10, fontweight='bold')

plt.xlim(0, 160)
plt.ylim(0, 180)
plt.title('Ejercicio 11: Región Factible y Combinación Óptima de Inventario', fontweight='bold')
plt.xlabel('Producto A (unidades, x)')
plt.ylabel('Producto B (unidades, y)')
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()
plt.tight_layout()
plt.show()

print("=" * 60)
print("EJERCICIO 11: RESULTADO")
print("=" * 60)
print("Punto de equilibrio óptimo: x = 25 unidades (Producto A), y = 125 unidades (Producto B)")
print("Presupuesto ejecutado: $50(25) + $30(125) = $5,000 USD")
print("Capacidad ocupada: 25 + 125 = 150 unidades\n")


# =========================================================
# EJERCICIO 12: MEDIA MÓVIL DE 3 DÍAS (SERIES DE TIEMPO)
# =========================================================
dias = np.arange(1, 11)
visitas = np.array([120, 140, 130, 150, 170, 160, 180, 200, 190, 210])

# Construcción del DataFrame
df_visitas = pd.DataFrame({'Día': dias, 'Visitas': visitas})
df_visitas['Media_Móvil_3D'] = df_visitas['Visitas'].rolling(window=3, center=True).mean()

print("=" * 60)
print("EJERCICIO 12: TABLA DE VISITAS Y MEDIA MÓVIL (3 DÍAS)")
print("=" * 60)
print(df_visitas.to_string(index=False))
print("\n")

# Gráfica de la Serie de Tiempo
plt.figure(figsize=(8, 4.5))
plt.plot(df_visitas['Día'], df_visitas['Visitas'], marker='o', label='Visitas Diarias (Original)', color='tab:blue', linewidth=2)
plt.plot(df_visitas['Día'], df_visitas['Media_Móvil_3D'], marker='s', linestyle='--', label='Media Móvil 3 Días', color='tab:red', linewidth=2)

plt.title('Ejercicio 12: Serie de Tiempo y Suavizado por Media Móvil', fontweight='bold')
plt.xlabel('Día')
plt.ylabel('Cantidad de Visitas')
plt.xticks(dias)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()
plt.tight_layout()
plt.show()


# =========================================================
# EJERCICIO 13: CLUSTERING K-MEANS 3D (SEGMENTACIÓN DE CLIENTES)
# =========================================================
# Matriz 3D: [Electrónicos, Ropa, Alimentos]
X_clientes = np.array([
    [50, 100, 200],  # Cliente 1
    [60, 90, 250],   # Cliente 2
    [70, 110, 300],  # Cliente 3
    [80, 120, 220],  # Cliente 4
    [90, 100, 270]   # Cliente 5
])

# Ajuste K-means con K = 2
kmeans = KMeans(n_clusters=2, random_state=42, n_init=10).fit(X_clientes)

print("=" * 60)
print("EJERCICIO 13: CLASSIFICACIÓN K-MEANS (3 CATEGORÍAS)")
print("=" * 60)
for i, (cliente, label) in enumerate(zip(X_clientes, kmeans.labels_), start=1):
    print(f"Cliente {i} {cliente} -> Pertenece al Grupo {label + 1}")

print("\nCentroides de los Grupos (Electrónicos, Ropa, Alimentos):")
for i, centroide in enumerate(kmeans.cluster_centers_, start=1):
    print(f"Grupo {i}: Electrónicos = ${centroide[0]:.2f} | Ropa = ${centroide[1]:.2f} | Alimentos = ${centroide[2]:.2f}")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats


# --- EJERCICIO 14 y 15 ---
# Datos del problema
X = np.array([2, 4, 6, 8, 10, 12])
Y = np.array([10, 20, 25, 35, 40, 50])

# Ajuste de regresión lineal
slope, intercept, r_value, p_value, std_err = stats.linregress(X, Y)

print("--- EJERCICIO 14 Y 15: REGRESIÓN Y CORRELACIÓN ---")
print(f"Pendiente (beta_1): {slope:.4f}")
print(f"Intercepto (beta_0): {intercept:.4f}")
print(f"Ecuación: Y = {slope:.4f}X + {intercept:.4f}")
print(f"Coeficiente de Correlación (r): {r_value:.4f}")
print(f"Coeficiente de Determinación (R^2): {r_value**2:.4f}")

# Gráfica de dispersión y línea de tendencia
plt.figure(figsize=(7, 5))
plt.scatter(X, Y, color='red', s=80, label='Datos Observados (Inversión vs Ventas)', zorder=5)
plt.plot(X, slope * X + intercept, color='blue', linestyle='-', label=f'Línea de Regresión: Y = {slope:.2f}X + {intercept:.2f}')

plt.title('Ejercicios 14 y 15: Regresión Lineal Simple y Ajuste de Datos')
plt.xlabel('Inversión en Publicidad (miles USD)')
plt.ylabel('Ventas (miles USD)')
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

print("=" * 65)
print("BLOQUE 6: VARIABILIDAD Y FUNCIONES APLICADAS (EJERCICIOS 16 A 20)")
print("=" * 65)

# -------------------------------------------------------------------------
# EJERCICIO 16: ANÁLISIS DE VARIABILIDAD EN INVENTARIOS (n = 8 semanas)
# -------------------------------------------------------------------------
inv = np.array([100, 120, 110, 130, 115, 125, 105, 140])

media_inv = np.mean(inv)
var_inv = np.var(inv, ddof=1)
std_inv = np.std(inv, ddof=1)
cv_inv = (std_inv / media_inv) * 100

print("\n--- EJERCICIO 16: INVENTARIOS ---")
print(f"Media Aritmética (x̄):     {media_inv:.2f} unidades")
print(f"Varianza Muestral (s²):   {var_inv:.2f} unidades²")
print(f"Desviación Estándar (s): {std_inv:.2f} unidades")
print(f"Coeficiente Variación (CV): {cv_inv:.2f}%")


# -------------------------------------------------------------------------
# EJERCICIO 17: EVALUACIÓN DE CALIFICACIONES (n = 10 estudiantes)
# -------------------------------------------------------------------------
calif = np.array([3.5, 4.0, 4.2, 3.8, 4.5, 2.8, 3.9, 4.1, 4.8, 3.4])

media_calif = np.mean(calif)
var_calif = np.var(calif, ddof=1)
std_calif = np.std(calif, ddof=1)
cv_calif = (std_calif / media_calif) * 100

print("\n--- EJERCICIO 17: CALIFICACIONES ---")
print(f"Media Aritmética (x̄):     {media_calif:.2f}")
print(f"Varianza Muestral (s²):   {var_calif:.4f}")
print(f"Desviación Estándar (s): {std_calif:.4f}")
print(f"Coeficiente Variación (CV): {cv_calif:.2f}%")


# -------------------------------------------------------------------------
# EJERCICIO 18: COMPARACIÓN DE CONSISTENCIA ENTRE EQUIPOS (A vs B)
# -------------------------------------------------------------------------
eq_A = np.array([80, 85, 90, 75, 80])
eq_B = np.array([60, 95, 100, 70, 85])

media_A, std_A = np.mean(eq_A), np.std(eq_A, ddof=1)
cv_A = (std_A / media_A) * 100

media_B, std_B = np.mean(eq_B), np.std(eq_B, ddof=1)
cv_B = (std_B / media_B) * 100

print("\n--- EJERCICIO 18: COMPARACIÓN DE EQUIPOS ---")
print(f"Equipo A -> Media: {media_A:.2f} | Desv. Est: {std_A:.2f} | CV: {cv_A:.2f}%")
print(f"Equipo B -> Media: {media_B:.2f} | Desv. Est: {std_B:.2f} | CV: {cv_B:.2f}%")
print("Conclusión: El Equipo A es más consistente y estable por tener menor CV.")


# -------------------------------------------------------------------------
# EJERCICIO 19: CONSUMO DE AGUA RESIDENCIAL (n = 10 hogares)
# -------------------------------------------------------------------------
agua = np.array([12, 15, 14, 18, 20, 16, 13, 17, 19, 15])

media_agua = np.mean(agua)
var_agua = np.var(agua, ddof=1)
std_agua = np.std(agua, ddof=1)
cv_agua = (std_agua / media_agua) * 100

print("\n--- EJERCICIO 19: CONSUMO DE AGUA ---")
print(f"Media Aritmética (x̄):     {media_agua:.2f} m³")
print(f"Varianza Muestral (s²):   {var_agua:.2f} m⁶")
print(f"Desviación Estándar (s): {std_agua:.2f} m³")
print(f"Coeficiente Variación (CV): {cv_agua:.2f}%")


# -------------------------------------------------------------------------
# EJERCICIO 20: FUNCIÓN LINEAL APLICADA (y = 15x + 50)
# -------------------------------------------------------------------------
# Ecuación de Costo Total: y = 15x + 50
f_costo = lambda x: 15 * x + 50

# Evaluación en los valores solicitados por la guía: x in {-2, -1, 0, 1, 2, 3}
x_eval = np.array([-2, -1, 0, 1, 2, 3])
y_eval = f_costo(x_eval)

df_ej20 = pd.DataFrame({
    'x (Unidades)': x_eval,
    'Procedimiento (15x + 50)': [f"15({x}) + 50" for x in x_eval],
    'y (Costo USD)': y_eval,
    'Par Ordenado (x, y)': [f"({x}, {y})" for x, y in zip(x_eval, y_eval)]
})

print("\n--- EJERCICIO 20: TABLA DE EVALUACIÓN DE LA FUNCIÓN LINEAL ---")
print(df_ej20.to_string(index=False))

print(f"\nPendiente (m): 15  -> Significado: Costo variable unitario ($15 por unidad adicional).")
print(f"Corte con eje y (b): 50 -> Significado: Costo fijo operativo en x = 0 ($50 USD).")

# Generación de la gráfica para el Ejercicio 20
x_linea = np.linspace(-3, 4, 100)
y_linea = f_costo(x_linea)

plt.figure(figsize=(8, 5))
plt.plot(x_linea, y_linea, color='purple', linewidth=2, label='Función: y = 15x + 50')
plt.scatter(x_eval, y_eval, color='red', s=80, zorder=5, label='Puntos evaluados (x, y)')

# Etiquetas para cada punto
for x_i, y_i in zip(x_eval, y_eval):
    plt.annotate(f'({x_i}, {y_i})', (x_i + 0.1, y_i - 2), fontsize=9)

plt.axhline(0, color='black', linewidth=0.8)
plt.axvline(0, color='black', linewidth=0.8)
plt.title('Ejercicio 20: Gráfica de la Función Lineal y Puntos Obtenidos', fontweight='bold')
plt.xlabel('x (Unidades)')
plt.ylabel('y (Costo Total USD)')
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()
plt.tight_layout()
plt.show()
```
