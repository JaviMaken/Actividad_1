## 2. Desarrollo (Paso 3): Aplicaciones Prácticas de la Ciencia de Datos y la Analítica de Datos en la ADRES

### 2.1 Contexto Institucional y Arquitectura Tecnológica de la ADRES
La ADRES administra los flujos financieros de más de 55.000 prestadores de salud en Colombia. Su plataforma analítica centraliza más de 110.645 millones de registros distribuidos en 21 catálogos de producción, operando sobre una infraestructura unificada en Microsoft Azure y Databricks que permite ejecutar consultas complejas sobre 100.000 millones de registros en menos de 10 segundos.

### 2.2 Aplicaciones de la Analítica de Datos (Business Intelligence y Diagnóstico)
La **analítica de datos** se enfoca en responder qué ocurrió y por qué ocurrió, transformando datos históricos en información estructurada para el control operacional:
1. **Tableros de Control de Giro Directo (Power BI):** Monitoreo continuo de más de 68.7 COP billones asignados a las Unidades de Pago por Capitación (UPC) en los regímenes Contributivo y Subsidiado.
2. **Auditaje de Inconsistencias por Contrastación:** Cruzamiento masivo entre la Base de Datos Única de Afiliados (BDUA) y los registros de servicios (RIPS/FEV-RIPS). Esta analítica diagnóstica identificó **471.480 personas fallecidas con servicios facturados posteriormente**, representando cobros inconsistentes por **COP $2.3 billones entre 2018 y 2023**.
3. **Seguimiento a Mandatos Judiciales:** Reportes de control para el seguimiento a los Autos 007, 089 y 504 de 2025 de la Corte Constitucional sobre medicamentos y presupuestos máximos.

### 2.3 Aplicaciones de la Ciencia de Datos (Modelación Predictiva e Inteligencia Artificial)
La **ciencia de datos** integra métodos estadísticos y modelos avanzados para descubrir patrones ocultos y proyectar comportamientos futuros:
1. **Modelos Econométricos de la UPC:** Proyección de la suficiencia financiera de la UPC y estimación del costo promedio por afiliado mediante modelos de regresión econométrica.
2. **Detección Automatizada de Anomalías:** Algoritmos supervisados y no supervisados diseñados para auditar la totalidad de las facturas electrónicas de venta (FEV-RIPS) sin requerir muestreos estadísticos parciales.
3. **Análisis de Concentración de Mercado:** Aplicación del Índice Herfindahl-Hirschman (HHI) para evaluar la estructura competitiva entre aseguradoras (EPS) en el periodo 2009–2024.
4. **Agentes de IA Conversacional:** Implementación de **Genie (Databricks)** y **Copilot (Power BI)** para la formulación de consultas en lenguaje natural mediante SQL automatizado.

```python
# Código de demostración analítica: Simulación de datos de Giro Directo ADRES
import pandas as pd
import numpy as np

# Semilla para reproducibilidad
np.random.seed(42)

# Crear dataset simulado de Giros Directos ADRES
regimenes = ['Contributivo', 'Subsidiado']
eps_list = ['EPS Sanitas', 'Sura EPS', 'Compensar', 'Nueva EPS', 'Salud Total']
departamentos = ['Bogotá D.C.', 'Antioquia', 'Valle del Cauca', 'Atlántico', 'Santander']

data = {
    'ID_Giro': range(1001, 1051),
    'Regimen': np.random.choice(regimenes, 50),
    'EPS': np.random.choice(eps_list, 50),
    'Departamento': np.random.choice(departamentos, 50),
    'Valor_Girado_COP': np.random.uniform(500e6, 12000e6, 50).round(2),
    'Estado_Afiliado': np.random.choice(['Activo', 'Activo', 'Activo', 'Fallecido_Reportado'], 50)
}

df_adres = pd.DataFrame(data)

# 1. Analítica Descriptiva: Resumen por Régimen y EPS
resumen_girado = df_adres.groupby(['Regimen', 'EPS'])['Valor_Girado_COP'].sum() / 1e9
print("=== RESUMEN DE GIROS DIRECTOS POR EPS Y RÉGIMEN (Miles de Millones COP) ===")
print(resumen_girado.round(2))

# 2. Analítica Diagnóstica: Detección de Giros a Afiliados Fallecidos
inconsistencias = df_adres[df_adres['Estado_Afiliado'] == 'Fallecido_Reportado']
total_inconsistente = inconsistencias['Valor_Girado_COP'].sum() / 1e6
print(f"\nInconsistencias detectadas: {len(inconsistencias)} registros por un total de COP{total_inconsistente:,.2f} Millones.")
```

## 3. Desarrollo (Paso 4): Tipos de Datos y Datasets en la Plataforma ADRES

### 3.1 Importancia de la Clasificación de los Datos
Saber identificar la estructura y el tipo de variable de los datos es fundamental en la ingeniería y ciencia de datos. Esta clasificación determina la elección del sistema de almacenamiento (relacional SQL vs. NoSQL / Data Lake), los métodos de limpieza aplicables y los algoritmos analíticos adecuados.

### 3.2 Clasificación por Grado de Estructura en la ADRES
1. **Datos Estructurados:** Información organizada rígidamente en tablas relacionales con esquemas predefinidos.
   * *Ejemplo ADRES:* **BDUA (Base de Datos Única de Afiliados)** con 16.913 millones de registros en 20 tablas relacionales, y la tabla de **Suficiencia UPC** (22.231 millones de registros).
2. **Datos Semiestructurados:** Archivos sin estructura tabular fija pero delimitados por etiquetas o claves organizativas.
   * *Ejemplo ADRES:* **FEV-RIPS (Factura Electrónica de Venta con RIPS integrado)** en formatos **XML y JSON**, provenientes de más de 55.000 prestadores con nodos anidados de atenciones médicas.
3. **Datos No Estructurados:** Archivos libres sin esquemas tabulares.
   * *Ejemplo ADRES:* Campos opcionales de **observaciones clínicas y diagnósticos secundarios** en texto libre dentro de los RIPS, requeridos de rutinas de Procesamiento de Lenguaje Natural (NLP).

### 3.3 Clasificación por Tipo de Variable
* **Variables Numéricas:**
  * *Discretas:* Número de atenciones médicas realizadas, días de estancia hospitalaria.
  * *Continuas:* Valor facturado en pesos colombianos (`Valor_Facturado_COP`), valor girado por la UPC.
* **Variables Categóricas:**
  * *Nominales:* Código de diagnóstico **CIE-10** (ej. J449), código de procedimiento **CUPS**, tipo de régimen (Contributivo vs. Subsidiado).
  * *Ordinales:* Nivel de complejidad de la IPS (Baja, Media, Alta complejidad).
* **Variables Temporales:**
  * Fechas de nacimiento, fechas de prestación de servicios y fechas de radicación. *Nota técnica:* Los sistemas heredados utilizaban fechas extremas como `1900-01-01` o `9999-12-31` para representar valores nulos, exigiendo reglas de depuración en la capa *Silver*.

```python
# Código de demostración de inspección de tipos de datos y depuración de fechas extremas (legados)
import pandas as pd

# Crear DataFrame que simula la recepción de RIPS con fechas extremas (nulos en sistemas antiguos)
data_rips = {
    'ID_Atencion': [2001, 2002, 2003, 2004, 2005],
    'Cod_CIE10': ['J449', 'I10X', 'E119', 'J449', 'Z000'],
    'Valor_COP': [150000.0, 450000.5, 1200000.0, 320000.0, 85000.0],
    'Nivel_IPS': ['Baja', 'Alta', 'Media', 'Baja', 'Media'],
    'Fecha_Servicio': ['2025-03-10', '1900-01-01', '2025-03-12', '9999-12-31', '2025-03-15']
}

df_rips = pd.DataFrame(data_rips)

print("=== TIPOS DE DATOS INICIALES (CAPA BRONZE) ===")
print(df_rips.dtypes)

# Depuración de Fechas Extremas (Estandarización Capa Silver)
df_rips['Fecha_Servicio_Limpia'] = pd.to_datetime(
    df_rips['Fecha_Servicio'].replace(['1900-01-01', '9999-12-31'], None)
)

print("\n=== DATOS PROCESADOS CON FECHAS LIMPIAS (CAPA SILVER) ===")
print(df_rips[['ID_Atencion', 'Fecha_Servicio', 'Fecha_Servicio_Limpia']])
```

## 4. Desarrollo (Paso 5): Ciclo de Vida de los Datos y Arquitectura Medallón en la ADRES

### 4.1 Etapas del Ciclo de Vida del Dato
1. **Captura y Recolección:** Ingesta distribuida desde 55.000 prestadores mediante conexiones VPN y enlaces directos con Microsoft Fabric OneLake.
2. **Almacenamiento:** Estructuración distribuida en Data Lakes en la nube de Azure con capacidad de escalamiento en Petabytes.
3. **Preparación y Limpieza:** Depuración de esquemas, eliminación de registros duplicados y conversión de estándares de codificación (CIE-9 a CIE-10).
4. **Análisis e Interpretación:** Consultas masivas en SQL Warehouses sobre Databricks y aplicación de modelos econométricos.
5. **Visualización y Comunicación:** Publicación de tableros interactivos en Power BI para el control gubernamental y la ciudadanía.
6. **Gobernanza y Conservación:** Control de acceso por roles (RBAC) y seguimiento del linaje del dato mediante **Unity Catalog**.

### 4.2 Integración con la Arquitectura Medallón (Azure / Databricks)
* **Capa Bronze (Raw Data):** Repositorio de datos crudos (JSON, XML e imágenes de facturas) almacenados tal como se reciben de las IPS.
* **Capa Silver (Cleansed Data):** Proceso de validación de reglas de negocio, conversión de tipos de datos, limpieza de fechas nulas y cruce relacional con la BDUA.
* **Capa Gold (Curated Business Data):** Tablas agregadas y denormalizadas preparadas para alimentar directamente los paneles de Power BI y los modelos de ciencia de datos.

```python
# Código de simulación de un Pipeline ETL bajo la Arquitectura Medallón (Bronze -> Silver -> Gold)
import pandas as pd

# 1. CAPA BRONZE: Ingesta de datos crudos
bronze_data = [
    {"id": "F001", "afiliado": "A12", "val": "150000", "f_nac": "1900-01-01"},
    {"id": "F002", "afiliado": "A15", "val": "350000", "f_nac": "1985-06-20"},
    {"id": "F003", "afiliado": "A12", "val": "220000", "f_nac": "1990-01-15"}
]

# 2. CAPA SILVER: Limpieza, tipado y transformación
silver_data = []
for row in bronze_data:
    val_clean = float(row["val"])
    f_nac_clean = None if row["f_nac"] == "1900-01-01" else row["f_nac"]
    silver_data.append({
        "id_factura": row["id"],
        "id_afiliado": row["afiliado"],
        "valor_cop": val_clean,
        "fecha_nacimiento": f_nac_clean
    })

df_silver = pd.DataFrame(silver_data)

# 3. CAPA GOLD: Agregación de métricas de negocio para la toma de decisiones
df_gold = df_silver.groupby("id_afiliado")["valor_cop"].agg(
    Total_Facturado_COP="sum",
    Promedio_Facturado_COP="mean"
).reset_index()

print("=== CAPA GOLD (MÉTRICAS CURADAS DE NEGOCIO PARA TABLEROS) ===")
print(df_gold)
```
